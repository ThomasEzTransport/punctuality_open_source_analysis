## Delays rate

For delays, we want nb of trains delays > 5 min at the terminus.
But how to identify the terminus in this dataset? It is always the last line of the Service_ID. 

```python
#Step 1: subset data with no cancellations (nor full, nor partial): 

intercity_wo_cancel = data_Intercity_2023[(data_Intercity_2023['Service:Completely cancelled'] == False)&(data_Intercity_2023['Service:Partly cancelled'] == False)]
len(intercity_wo_cancel['Service_ID'].unique())
```

```python
#Step 2: add a new index
delays_intercity_2023 = intercity_wo_cancel
delays_intercity_2023['Index'] = range(1, len(delays_intercity_2023) + 1)
delays_intercity_2023.head(5)
```

```python
#the filtering #3 should not be here, but after step #4
```

```python
#4 for each Service_ID select the maximum index, to have only the terminus
delays_terminus_intercity_2023 = []
unique_Service_ID = delays_intercity_2023['Service_ID'].unique()

for Id in unique_Service_ID: 
    subset_temp = delays_intercity_2023[delays_intercity_2023['Service_ID'] == Id]
    delay_terminus = subset_temp[subset_temp['Index'] == max(subset_temp['Index'])]
    delays_terminus_intercity_2023.append(delay_terminus)
```

```python
delays_terminus_intercity_2023 = pd.concat(delays_terminus_intercity_2023, ignore_index=True)
```

```python
#3 To lighten the dataset, already get rid off travel with no delays
delays_terminus_intercity_2023 = delays_terminus_intercity_2023[delays_terminus_intercity_2023["Stop:Arrival delay"] > 5]
```

```python
delays_terminus_intercity_2023_5min = delays_terminus_intercity_2023[delays_terminus_intercity_2023['Stop:Arrival delay'] > 5]
delays_terminus_intercity_2023_10min = delays_terminus_intercity_2023[delays_terminus_intercity_2023['Stop:Arrival delay'] > 10]
delays_terminus_intercity_2023_15min = delays_terminus_intercity_2023[delays_terminus_intercity_2023['Stop:Arrival delay'] > 15]
```

```python
nb_Intercity_running_2023 = total_Intercity_2023 - full_cancelled_Intercity_2023 - part_cancelled_Intercity_2023
```

```python
punctual_terminus_intercity_2023_5min = nb_Intercity_running_2023 - len(delays_terminus_intercity_2023_5min)
punctual_terminus_intercity_2023_10min = nb_Intercity_running_2023 - len(delays_terminus_intercity_2023_10min)
punctual_terminus_intercity_2023_15min = nb_Intercity_running_2023 - len(delays_terminus_intercity_2023_15min)
```

```python
punctuality_rate_5min = (punctual_terminus_intercity_2023_5min/nb_Intercity_running_2023)*100
punctuality_rate_10min = (punctual_terminus_intercity_2023_10min/nb_Intercity_running_2023)*100
punctuality_rate_15min = (punctual_terminus_intercity_2023_15min/nb_Intercity_running_2023)*100
punctuality_rate_5min
```

```python
(total_Intercity_2023 - len(delays_terminus_intercity_2023_5min))/total_Intercity_2023
```

```python
print(f'In 2023, their were {nb_Intercity_running_2023} Intercity trains running')
print(f'Among which {round(punctuality_rate_5min,2)}% were punctual (delay < 5 min)')
```
